```python
# import pandas library for data manipulation and analysis
import pandas as pd
```

```python
# reading the csv files into separate dataframes
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv', low_memory=False)
```

```python
# select columns
df_movies = df_movies[['imdb_title_id', 'title', 'year', 'genre', 'country']]
df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

```python
# display the dataframe df_movies
df_movies
```

```python
# display the dataframe df_ratings
df_ratings
```

## **merge()**

```python
# create first dataframe with id and age columns
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                   'age': [30, 23, 25, 22]})

# create second dataframe with id and job column
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'],
                    'job': ['Doctor', 'Statistician', 'Accountant', 'Developer']})
```

```python
# show dataframe df1
df1
```

```python
# show dataframe df1
df2
```

## **Inner Join : (A ∩ B)**

```python
# merge df1 and df2 on 'id' using inner join
df1_merge_inner_df2 = df1.merge(df2, how='inner', on='id')
```

```python
# merged dataframe using inner join
df1_merge_inner_df2
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1_merge_inner_df2.shape)
```

```python
# merge df_movies and df_ratings (inner join)
df_inner_join = df_movies.merge(df_ratings, how='inner', on='imdb_title_id')
```

```python
# show merged dataframe using inner join
df_inner_join
```

```python
# shape
print(df_movies.shape)
print(df_ratings.shape)
print(df_inner_join.shape)
```

## **Outer Join (Full Join) : (A ∪ B)**

```python
# merge df1 and df2 on 'id' using outer join
df1_merge_outer_df2 = df1.merge(df2, how='outer', on='id')
```

```python
# merged dataframe using outer join
df1_merge_outer_df2
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1_merge_outer_df2.shape)
```

```python
# merge df_movies and df_ratings (outer join)
df_outer_join = df_movies.merge(df_ratings, how='outer', on='imdb_title_id')
```

```python
# show merged dataframe using outer join
df_outer_join
```

```python
# shape
print(df_movies.shape)
print(df_ratings.shape)
print(df_outer_join.shape)
```

## **Exclusive Outer Join (Exclusive Full Join) : (A ∪ B) – (A ∩ B)**

```python
# outer join
df1_merge_outer_df2 = df1.merge(df2, how='outer', on='id')
df1_merge_outer_df2
```

```python
# indicator=True
df1_merge_outer_df2 = df1.merge(df2, how='outer', on='id', indicator=True)
df1_merge_outer_df2
```

```python
# query (for exclusive outer join)
df1_merge_exclusive_outer_df2 = df1_merge_outer_df2.query('_merge == "left_only" or _merge == "right_only"')
df1_merge_exclusive_outer_df2
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1_merge_exclusive_outer_df2.shape)
```

```python
# outer join
df_outer_join = df_movies.merge(df_ratings, how='outer', on='imdb_title_id')
df_outer_join
```

```python
# indicator=True
df_outer_join = df_movies.merge(df_ratings, how='outer', on='imdb_title_id', indicator=True)
df_outer_join
```

```python
# query (for exclusive outer join)
df_exclusive_outer_join = df_outer_join.query('_merge == "left_only" or _merge == "right_only"')
df_exclusive_outer_join
```

```python
# shape
print(df_movies.shape)
print(df_ratings.shape)
print(df_exclusive_outer_join.shape)
```
