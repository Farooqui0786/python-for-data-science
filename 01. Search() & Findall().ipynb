## **The re module**

```python
# import regular expressions module
import re
```

```python
text = """Hello World.
I love Python!!!!!
My first lucky number is 987-654-321
My second lucky number is 876-543-210
My third lucky number is 765-432-100
My fourth lucky number is 123-456-123-123-456
"""
```

```python
date = '''
13-04-2021
2021-13-04
2021-04-13
'''
```

```python
char = '''
username10
abc
10
'''
```

### **re.search()**

```python
# return the first digit found in text
re.search(r'\d', text)
```

```python
# flags: search for 'hello' at the start of text (case-insensitive)
re.search(r'^hello', text, flags=re.I)
```

### **re.findall()**

```python
# return all digits found in text
re.findall(r'\d', text)
```

```python
# match only punctuation: all special characters (non-alphanumeric, non-space) in text
re.findall(r'[^\w\s]', text)
```

```python
# find all dates in the format (MM-DD-YYYY) in date
re.findall(r'\d{2}-\d{2}-\d{4}', date)
```

```python
# match right username format (4-14 characters and only letters or digits)
re.findall(r'[a-zA-Z0-9]{4,14}', char)
```
