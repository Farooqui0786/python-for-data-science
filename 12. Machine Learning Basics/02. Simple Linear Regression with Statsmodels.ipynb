```python
# import pandas for data handling
import pandas as pd
```

```python
# read the csv file into dataframe & show the dataset
df_boston = pd.read_csv('Boston House Prices.csv')
df_boston
```

## **Simple Linear Regression with Statsmodels**

```python
# import statsmodels library for statistical modeling
import statsmodels.api as sm
```

```python
# define dependent and independent variable
y = df_boston['Value']      # dependent
x = df_boston['Rooms']      # independent
```

```python
# exploring the dataset with scatter plot
df_boston.plot(kind='scatter', x='Rooms', y='Value', figsize=(9, 6))
```

### **Making the Regression: Adding a constant and fitting the model**

**Statsmodels doesn't add a constant by default, so we have to do it on our own**

```python
x = sm.add_constant(x)      # adding constant
lm = sm.OLS(y,x).fit()      # fitting the model
```

```python
# use the trained model 'lm' to make predictions on input data x
lm.predict(x)
```

```python
# the regression table
lm.summary()
```

### **Linear Regression Equation**

```python
# rooms coef: 9.1021
# constant coef: -34.6706

# linear equation: y = ax + b
y_pred = 9.1021 * x['Rooms'] - 34.6706
```

### **Plotting the Regression line**

```python
# import seaborn for statistical data visualization
import seaborn as sns

# import matplotlib for general plotting
import matplotlib.pyplot as plt

# plotting the datapoints
plt.figure(figsize=(9, 6), tight_layout=True)
sns.scatterplot(x=x['Rooms'], y=y)

# plotting the regression line
sns.lineplot(x=x['Rooms'], y=y_pred, color='red')
plt.show()
```
