```python
# import pandas library for data manipulation and analysis
import pandas as pd
```

```python
# reading the csv files into separate dataframes
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv', low_memory=False)
```

```python
# select columns
df_movies = df_movies[['imdb_title_id', 'title', 'year', 'genre', 'country', 'director', 'actors']]
df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

```python
# display the dataframe df_movies
df_movies
```

```python
# display the dataframe df_ratings
df_ratings
```

## **concat()**

### **Concatenate Vertically**

To concatenate vertically (along hte rows) we should have columns in common between the 2 dataframes

```python
# create first dataframe with id and age columns
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                   'age': [30, 23, 25, 22]})

# create second dataframe with id and age columns
df2 = pd.DataFrame({'id': ['E', 'F', 'G', 'H'],
                   'age': [40, 21, 19, 24]})
```

```python
# concatenate df1 and df2 into a single dataframe (vertically)
pd.concat([df1, df2])
```

```python
# extract a 50% sample of the original dataframe
df_sample = df_movies.sample(frac=0.5)

# show dataframe
df_sample
```

```python
# shape of the dataframe that we'll concatenate
print(df_movies.shape)
print(df_sample.shape)
```

```python
# concatenate df_movies and df_sample (vertically along the rows)
df_concat_vertically = pd.concat([df_movies, df_sample])

# show dataframe
df_concat_vertically
```

```python
# shape of the concatenated dataframe
df_concat_vertically.shape
```

### **Concatenate Horizontally**

To concatenate horizontally (along the columns) we should have a common index between the 2 dataframes

```python
# create first dataframe with id and age columns
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                   'age': [30, 23, 25, 22]})

# create second dataframe with job column
df2 = pd.DataFrame({'job': ['Doctor', 'Statistician', 'Accountant', 'Developer']})
```

```python
# concatenate df1 and df2 into a single dataframe (horizontally)
pd.concat([df1, df2], axis=1)
```

```python
# sahpe of dataframe
print(df_movies.shape)
print(df_ratings.shape)
```

```python
# set imdb_title_id as index for movies and ratings dataframe
df_movies = df_movies.set_index('imdb_title_id')
df_ratings = df_ratings.set_index('imdb_title_id')
```

```python
# sahpe of dataframe that we'll concatenate
print(df_movies.shape)
print(df_ratings.shape)
```

```python
# concatenate df_movies and df_ratings on imdb_title_id (horizontally along the columns)
df_concat_horizontally = pd.concat([df_movies, df_ratings], axis=1)

# show dataframe
df_concat_horizontally
```

```python
# shape of the concatenated dataframe
df_concat_horizontally.shape
```
